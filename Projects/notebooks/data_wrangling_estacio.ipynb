```python
canais = {
    'SMS': df_sms,
    'RCS': df_rcs,
    'WHATS': df_whatsapp,  # remove se não houver dados
    'EMAIL': df_email
}

import pandas as pd
import time

def preparar_canal(df: pd.DataFrame, canal: str) -> pd.DataFrame:
    df = df.copy()
    df['CPF'] = df['CPF'].str.strip()

    if canal in ['SMS', 'RCS', 'WHATS']:
        df = df[['DATA_DISPARO', 'CPF', 'CUSTO', 'TELEFONE']].rename(columns={'TELEFONE': 'CONTATO'})

    elif canal == 'EMAIL':
        df = df[['DATA', 'CPF', 'CUSTO', 'EMAIL']].rename(columns={
            'DATA': 'DATA_DISPARO',
            'EMAIL': 'CONTATO'
        })

    df['CANAL'] = canal
    return df


def cruzar_mailing_canal(df_mailing: pd.DataFrame, df_canal: pd.DataFrame, nome_canal: str) -> tuple[pd.DataFrame, pd.DataFrame]:
    
    inicio = time.time()
    print(f"\n[{nome_canal}] Iniciando cruzamento...")

    # 1. Prepara o canal
    df_canal_prep = preparar_canal(df_canal, nome_canal)

    # 2. Remove espaços do mailing
    df_mailing['CPF'] = df_mailing['CPF'].str.strip()

    # 3. Filtra mailing pelos CPFs do canal
    cpfs_canal = df_canal_prep['CPF'].unique()
    df_mailing_filtrado = df_mailing[df_mailing['CPF'].isin(cpfs_canal)]
    print(f"[{nome_canal}] Mailing filtrado: {df_mailing_filtrado.shape[0]} linhas")

    # 4. Deduplica o mailing
    df_mailing_dedup = (
        df_mailing_filtrado
        .assign(prioridade_grupo=df_mailing_filtrado['GRUPO'].apply(lambda x: 1 if x == 'ESTACIO' else 2))
        .sort_values(['CPF', 'DATA', 'prioridade_grupo', 'ATRASO', 'CONTRATO'],
                     ascending=[True, True, True, False, True])
        .drop_duplicates(subset=['CPF', 'DATA'], keep='first')
        .drop(columns='prioridade_grupo')
    )
    print(f"[{nome_canal}] Mailing deduplicado: {df_mailing_dedup.shape[0]} linhas")

    # 5. Merge com indicador para identificar o que ficou de fora
    df_merge = df_canal_prep.merge(
        df_mailing_dedup,
        left_on=['CPF', 'DATA_DISPARO'],
        right_on=['CPF', 'DATA'],
        how='left',
        indicator=True
    )

    # 6. Separa os dois dfs
    df_com_relacionamento = df_merge[df_merge['_merge'] == 'both'].drop(columns='_merge')
    df_sem_relacionamento = df_merge[df_merge['_merge'] == 'left_only'][['DATA_DISPARO', 'CPF', 'CUSTO', 'CONTATO', 'CANAL']]

    tempo = time.time() - inicio
    print(f"[{nome_canal}] Concluído em {tempo:.1f}s")
    print(f"[{nome_canal}] Com relacionamento: {df_com_relacionamento.shape[0]} linhas")
    print(f"[{nome_canal}] Sem relacionamento: {df_sem_relacionamento.shape[0]} linhas — custo fora: R$ {df_sem_relacionamento['CUSTO'].sum():,.2f}")

    return df_com_relacionamento, df_sem_relacionamento


def cruzar_todos_canais(df_mailing: pd.DataFrame, canais: dict) -> tuple[pd.DataFrame, pd.DataFrame]:
    resultados_com = []
    resultados_sem = []
    inicio_total = time.time()

    for nome_canal, df_canal in canais.items():
        df_com, df_sem = cruzar_mailing_canal(df_mailing, df_canal, nome_canal)
        resultados_com.append(df_com)
        resultados_sem.append(df_sem)

    df_consolidado_com = pd.concat(resultados_com, ignore_index=True)
    df_consolidado_sem = pd.concat(resultados_sem, ignore_index=True)

    tempo_total = time.time() - inicio_total
    print(f"\nTotal com relacionamento: {df_consolidado_com.shape[0]} linhas")
    print(f"Total sem relacionamento: {df_consolidado_sem.shape[0]} linhas — custo total fora: R$ {df_consolidado_sem['CUSTO'].sum():,.2f}")
    print(f"Tempo total: {tempo_total:.1f}s")

    return df_consolidado_com, df_consolidado_sem

#df_consolidado = cruzar_todos_canais(df_mailing_hist, canais)
df_com_relacionamento, df_sem_relacionamento = cruzar_todos_canais(df_base_processing, canais)
```

```python
import os
from pathlib import Path

BASE_PATH = Path(r"C:\Users\claudiano.alves\Documents\Claudiano\repository\Data-Analysis-Portfolio\Projects\Estacio\process_digital_channels\Data\data_analytcs")

def salvar_analiticos(df_com: pd.DataFrame, df_sem: pd.DataFrame, canal: str) -> None:
    """
    Salva os analíticos de cada canal separados por relacionamento com mailing,
    ano, mês e dia.
    
    Parâmetros:
        df_com: df com relacionamento com mailing
        df_sem: df sem relacionamento com mailing
        canal: nome do canal (SMS, RCS, WHATS, EMAIL)
    """
    canal_lower = canal.lower()
    pasta_canal = BASE_PATH / f"analytical_{canal_lower}"

    dfs = {
        'com_relacion_mailing': df_com,
        'sem_relacion_mailing': df_sem
    }

    for tipo, df in dfs.items():
        if df.empty:
            print(f"[{canal}] {tipo} — df vazio, pulando...")
            continue

        # Garante que DATA_DISPARO é datetime
        df = df.copy()
        df['DATA_DISPARO'] = pd.to_datetime(df['DATA_DISPARO'])

        # Itera por data única
        for data, df_dia in df.groupby(df['DATA_DISPARO'].dt.date):
            ano = str(data.year)
            mes = f"{data.month:02d}"
            dia = f"{data.month:02d}{data.day:02d}"  # ex: 0501

            # Cria estrutura de pastas se não existir
            pasta_destino = pasta_canal / tipo / ano / mes
            pasta_destino.mkdir(parents=True, exist_ok=True)

            # Nome do arquivo ex: analitico_email_0501.csv
            nome_arquivo = f"analitico_{canal_lower}_{dia}.csv"
            caminho_arquivo = pasta_destino / nome_arquivo

            df_dia.to_csv(caminho_arquivo, index=False, sep=';', encoding='utf-8-sig')
            print(f"[{canal}] {tipo} — salvo: {caminho_arquivo}")


def salvar_todos_analiticos(df_com: pd.DataFrame, df_sem: pd.DataFrame) -> None:
    """
    Salva os analíticos de todos os canais presentes nos dfs consolidados.
    """
    inicio = time.time()

    for canal in df_com['CANAL'].unique():
        salvar_analiticos(
            df_com=df_com[df_com['CANAL'] == canal],
            df_sem=df_sem[df_sem['CANAL'] == canal],
            canal=canal
        )

    tempo = time.time() - inicio
    print(f"\nTodos os analíticos salvos em {tempo:.1f}s")

salvar_todos_analiticos(df_com_relacionamento, df_sem_relacionamento)
```
